# Práctica 01 para series de tiempo

Para esta práctica, se utilizará la librería `yfinance`, la cual permite obtener datos históricos de mercados financieros directamente desde Yahoo Finance. Esta herramienta es de uso gratuito y no requiere el uso de claves de acceso (API keys), lo que la hace especialmente conveniente para fines educativos.

Los datos obtenidos mediante `yfinance` corresponden a información real del mercado financiero. Sin embargo, es importante considerar que esta herramienta no constituye una fuente oficial de datos, por lo que puede presentar limitaciones en cuanto a disponibilidad, consistencia o actualización. Para fines académicos y de análisis exploratorio, estos datos son adecuados; no obstante, para aplicaciones profesionales o decisiones financieras, se recomienda el uso de fuentes oficiales o servicios especializados.

Adicionalmente, `yfinance` no es una API oficial, por lo que en ocasiones puede presentar fallas temporales en la descarga de datos.

Los principales argumentos que requiere yfinance para realizar la descarga son:

+ `tickers`: Los tickers son identificadores únicos de activos financieros (por ejemplo, acciones) en los mercados. Por ejemplo, `AAPL` corresponde a Apple, `TSLA` a Tesla y `MSFT` a Microsoft. Si quieres inspeccionar tickers [pulsa aquí](https://www.nasdaq.com/market-activity/stocks/screener) y se te abrirá un catálogo de diversas multinacionales.
+ `start`: Fecha de inicio para la descarga de datos en formato `YYYY-mm-dd` que significa: cuatro dígitos para el año (`YYYY`), dos dígitos para el mes (`mm`) y dos dígitos para el día (`dd`) en ese orden.
+ `end`: Fecha de fin para la descarga de datos siguiendo también un formato `YYYY-mm-dd`.
+ `interval`: Define la frecuencia de los datos. Por ejemplo, `"1d"` indica datos diarios, `"1wk"` semanales y `"1mo"` mensuales.
+ `group_by`: Define la forma en que se organizan las columnas cuando se descargan datos de uno o varios tickers. Por ejemplo, `group_by="column"` organiza la información por tipo de variable (Open, High, Low, Close, etc.), mientras que `group_by="ticker"` agrupa primero por activo financiero.
+ `auto_adjust`: Si se establece en `True`, los precios se ajustan automáticamente considerando dividendos y divisiones de acciones. Si es `False`, se mantienen los valores originales.
+ `actions`: Permite incluir información adicional como dividendos y divisiones de acciones cuando se establece en `True`.
+ `progress`: Controla si se muestra o no una barra de progreso durante la descarga.

### Análisis de la Tendencia con Suavización Exponencial Simple (SES - Span 252)
La Suavización Exponencial Simple (SES) con un parámetro de span de 252 (aproximadamente un año bursátil) nos permite modelar de forma óptima la inercia a largo plazo:
* **Memoria de Datos:** A diferencia del promedio móvil clásico, este método asigna ponderaciones que disminuyen de forma exponencial, reteniendo influencia de todos los datos históricos (incluyendo los más antiguos).
* **Comportamiento:** La curva verde resultante es sumamente suave y libre de ruido. Muestra de manera robusta cómo Walmart se ha mantenido en una clara e indiscutible tendencia de crecimiento sostenido a largo plazo, sin verse afectada de manera brusca por correcciones locales menores.

### Ejercicio 3: Comparativa de Métodos (SMA y SES)

Tras analizar los comportamientos de las curvas del **Promedio Móvil Simple (SMA - 66 días)** y la **Suavización Exponencial Simple (SES - Span 252)** sobre la serie de tiempo de Walmart, se identifican las siguientes ventajas y desventajas:

#### **1. Promedio Móvil Simple (SMA)**
* **Ventajas:**
  * **Fácil interpretación:** Representa de manera muy intuitiva el promedio real de los precios en el periodo seleccionado (en este caso, el trimestre).
  * **Detección de puntos de inflexión:** Es más sensible a cambios de tendencia de mediano plazo, permitiendo observar el aplanamiento y descenso de la curva hacia mediados de 2026 de forma más oportuna.
* **Desventajas:**
  * **Efecto de desfase (Lag):** Al ponderar por igual todos los días, la señal reacciona con retraso a los movimientos abruptos del precio.
  * **Pérdida de datos iniciales:** Requiere acumular $N$ observaciones (66 días) antes de poder graficar el primer punto, dejando un vacío visual al inicio del histórico.
  * **Falta de memoria histórica total:** Ignora por completo cualquier dato anterior a la ventana de 66 días.

#### **2. Suavización Exponencial Simple (SES)**
* **Ventajas:**
  * **Memoria infinita ponderada:** No desecha datos antiguos; en su lugar, les asigna un peso decreciente exponencialmente, permitiendo que toda la historia del activo influya en la curva actual.
  * **Disponibilidad inmediata:** Se puede calcular desde el primer día de la serie de tiempo sin necesidad de esperar a acumular una ventana de datos.
  * **Suavizado superior a largo plazo:** Al configurarse con un span amplio (252 días), elimina por completo el ruido y muestra una inercia de crecimiento sumamente robusta y libre de oscilaciones transitorias.
* **Desventajas:**
  * **Menor sensibilidad a corto/mediano plazo:** Debido a su alta inercia (span de 252), ignora la desaceleración o corrección hacia la baja que se presenta a mediados de 2026, lo que podría retrasar la toma de decisiones si se busca detectar cambios de ciclo recientes.

### Análisis de la Tendencia Polinomial (Regresión de Grado 2)
Al aplicar un ajuste polinomial de segundo grado sobre los precios de cierre ajustados de Walmart, logramos filtrar las oscilaciones diarias para identificar la trayectoria principal:
* **Dirección general:** Se aprecia un crecimiento sostenido desde inicios de 2025, el cual se acelera a partir de la segunda mitad del año.
* **Comportamiento:** La curva cian (Tendencia) suaviza perfectamente las caídas de corto plazo (como la observada entre marzo y abril de 2025), confirmando que el activo se encuentra bajo un fuerte ciclo alcista generalizado.

```python
# Aquí no le muevan a nada
# Básicamente, lo que se hace este bloque es crear la función
# Transformada_de_Fourier, puesto a que no existe para lo que la quiero usar.

def Transformada_de_Fourier(serie, terminos, un_grafico):
  '''La función acepta los argumentos "serie", la cual debe de ser un array
  en numpy y en pandas de la forma df["my_variable"].values. Por otra parte,
  el parámetro "terminos" es un número natural que indica la cantidad de
  términos que desarrolla la serie. Entre mayor sea el número de términos,
  la serie será más precisa, pero más difícil de interpretar.

  La función tiene como salida una lista con los componentes sinosoidales de serie.
  El último elemento de la lista, es la suma de todos los componentes'''

  import numpy as np
  import matplotlib.pyplot as plt
  import pandas as pd
  from scipy.fftpack import fft, ifft

  # Supongamos que tienes una serie en df['AAPL_Close']
  # Reemplaza esta línea con tu DataFrame y serie específicos
  n = len(serie)

  # Calcular la transformada de Fourier
  transformada_fourier = fft(serie)

  # Obtener las frecuencias
  frecuencias = np.fft.fftfreq(n)

  # Crear un DataFrame para almacenar la frecuencia y su magnitud
  componentes = pd.DataFrame({
      'frecuencia': frecuencias,
      'magnitud': np.abs(transformada_fourier),
      'longitud_de_onda': 1 / frecuencias
  })

  # Ordenar el dataframe de mayor a menor en términos de magnitud

  componentes = componentes.sort_values(by='magnitud', ascending=False)

  # Seleccionar las frecuencias con mayor magnitud (excluyendo la frecuencia cero)
  top_frecuencias = componentes.loc[componentes['frecuencia'] > 0].nlargest(terminos, 'magnitud')
  top_frecuencias.reset_index(drop=True, inplace=True)
  print("Frecuencias principales:\n", top_frecuencias)

  # Crear el índice de tiempo para la serie
  t = np.arange(n)

  # Graficar cada componente de frecuencia junto con la serie original
  plt.figure(figsize=(12, 4))

  componente_temporal_sumado = np.zeros_like(serie)
  componentes_temporales = []

  n=0
  for i, row in top_frecuencias.iterrows():
      n+=1
      # Copiar la transformada de Fourier y mantener solo la frecuencia actual
      fourier_component = np.zeros_like(transformada_fourier)
      idx = np.where(frecuencias == row['frecuencia'])[0][0]  # índice de la frecuencia en la FFT
      fourier_component[idx] = transformada_fourier[idx]  # mantener solo la frecuencia positiva
      fourier_component[-idx] = transformada_fourier[-idx]  # mantener la frecuencia negativa correspondiente
      if n == (terminos+1):
        break

      # Reconstruir la señal en el tiempo
      componente_temporal = ifft(fourier_component).real
      componentes_temporales.append(componente_temporal)
      componente_temporal_sumado += componente_temporal

      # Graficar la componente
      plt.plot(
          componente_temporal,
          label=f'Longitud de onda {1 / row["frecuencia"]:.0f}',
          alpha=1,
          linewidth = 0.5,
          )
      plt.title('Componentes de Fourier de la Serie')
      plt.xlabel('Tiempo')
      plt.ylabel('Valor')
      plt.legend()
      plt.grid()

  componentes_temporales.append(componente_temporal_sumado)
  plt.plot(serie, label='Serie Original', color='black', alpha=0.5)
  if not un_grafico:
    plt.figure(figsize=(12, 4))
    plt.plot(serie, label='Serie Original', color='black', alpha=0.5)
  plt.plot(componente_temporal_sumado, label='Componente temporal sumada', color='red')
  plt.legend()
  plt.title('Suma de los Componentes de Fourier de la Serie')
  plt.xlabel('Tiempo')
  plt.ylabel('Valor')
  plt.grid()
  plt.show()

  return componente_temporal_sumado
```

```python
serie = df['ruido_estacionalidad_polinomial'].values
terminos = 5
un_grafico = False

cts = Transformada_de_Fourier(serie, terminos, un_grafico)
# cts: componente temporal sumado

df["estacionalidad_polinomial"] = cts
```

#### **1. Análisis de las Longitudes de Onda (Estacionalidad)**
El algoritmo identificó las 5 componentes cíclicas o frecuencias con mayor magnitud en la señal residual (estacionaria). Considerando que un mes bursátil tiene aproximadamente **21 días hábiles**, analizamos el significado de cada longitud de onda obtenida:
* **Longitud de onda de 88.00 observaciones:** Equivale aproximadamente a **4.2 meses** (un cuatrimestre). Sugiere la presencia de un ciclo estacional fuerte asociado con la presentación de reportes financieros cuatrimestrales o ciclos de reabastecimiento e inventario estacional.
* **Longitud de onda de 70.40 observaciones:** Representa cerca de **3.3 meses**, lo cual se alinea firmemente con el comportamiento estacional de carácter trimestral típico del sector minorista.

**Conclusión de Estacionalidad:** Se evidencia una clara dominancia de ciclos de mediano plazo (trimestrales y bimensuales) en las acciones de Walmart, en lugar de patrones de muy corto plazo (semanales o mensuales puros). Esto es consistente con una empresa multinacional cuyo valor responde de manera cíclica a las dinámicas de consumo trimestral y anuncios corporativos periódicos.

El análisis muestra que el ruido fluctúa de manera estacionaria alrededor de cero sin exhibir patrones sistemáticos visuales, lo que valida que la descomposición logró capturar exitosamente la estructura subyacente de la serie.

### Ejercicio 6: Análisis Fundamental (Ligero) de los Componentes de la Serie

A continuación, se presenta una interpretación razonada y contextualizada del comportamiento financiero de **Walmart (WMT)** entre enero de 2025 y junio de 2026, vinculando los hallazgos matemáticos con el entorno de negocio y las dinámicas reales del mercado retail:

#### **1. Explicación de la Tendencia (Largo Plazo)**
La tendencia de Walmart muestra un crecimiento constante durante 2025 que parece acelerarse hacia la segunda mitad del año, seguido de una estabilización e inflexión moderada a mediados de 2026. Los factores fundamentales que podrían justificar este comportamiento incluyen:
* **Estrategia Omnicanal y Comercio Electrónico:** Durante los últimos años, Walmart ha consolidado su infraestructura de e-commerce y servicios de suscripción (*Walmart+*), compitiendo directamente con gigantes tecnológicos. El crecimiento de la tendencia refleja el éxito de esta transición digital y la captura de márgenes de ganancia más amplios.
* **Defensiva ante la Inflación:** Walmart es conocido por su política de *"Precios Bajos Todos los Días"*. En periodos de incertidumbre económica o inflación moderada, los consumidores tienden a migrar de tiendas premium hacia Walmart para optimizar su presupuesto diario, lo que impulsa las ventas de la compañía de manera contracíclica.
* **Inflexión a mediados de 2026:** El aplanamiento de la curva hacia mediados de 2026 puede explicarse por una estabilización general de las tasas de interés, saturación temporal de sus nuevos canales de distribución o un enfriamiento general en el consumo minorista global.

#### **2. Justificación de la Estacionalidad (Mediano Plazo)**
La descomposición de Fourier reveló ciclos estacionales dominantes de carácter trimestral y bimonthly (longitudes de onda de ~88 y ~70 días hábiles). Estos patrones cíclicos responden directamente a hitos operativos clave del sector minorista:
* **Reportes Financieros Trimestrales:** Cada tres meses, Walmart presenta de manera oficial sus estados financieros. Las expectativas de los analistas e inversores generan movimientos cíclicos recurrentes de acumulación o distribución de acciones alrededor de estas fechas clave.
* **Ciclos de Consumo Mayor y Reabastecimiento:** Existen eventos anuales de consumo masivo que dictan el flujo de caja, tales como el inicio del año escolar (*Back to School* en el tercer trimestre), la temporada navideña (*Holiday Season* en el cuarto trimestre) y las liquidaciones de inventario de primavera. El mercado anticipa y descuenta estas fluctuaciones trimestralmente en la cotización.

#### **3. Interpretación del Ruido de Medición (Corto Plazo)**
El ruido aislado representa la volatilidad diaria no predecible mediante ciclos ni curvas matemáticas de largo plazo. Este comportamiento aleatorio se atribuye a factores coyunturales del día a día, tales como:
* **Noticias Macroeconómicas:** Publicaciones diarias de datos de empleo, inflación (IPC) o decisiones de la Reserva Federal (Fed) que causan fluctuaciones inmediatas y momentáneas en todo el mercado accionario.
* **Eventos Geopolíticos y Logísticos:** Interrupciones repentinas en las cadenas de suministro globales, fluctuaciones en los precios de los combustibles (que afectan directamente los costos de distribución terrestre de Walmart) o noticias corporativas inesperadas.
* **Volatilidad de Sentimiento:** Transacciones masivas intradía generadas por algoritmos de trading de alta frecuencia que buscan liquidez inmediata, creando pequeñas fluctuaciones aleatorias que se revierten en el corto plazo.
